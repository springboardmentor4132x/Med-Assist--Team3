```python
import pandas as pd
import numpy as np

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# Load the shared train and test datasets
train_df = pd.read_csv("../dataset/splits/train.csv")
test_df = pd.read_csv("../dataset/splits/test.csv")

print("Training dataset:", train_df.shape)
print("Testing dataset:", test_df.shape)

display(train_df.head())
```

```python
# Separate features and target
X_train = train_df.drop(columns=["Disease"])
y_train = train_df["Disease"]

X_test = test_df.drop(columns=["Disease"])
y_test = test_df["Disease"]

# Verify feature consistency
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)

print("\nTraining target classes:", y_train.nunique())
print("Testing target classes:", y_test.nunique())

print("\nFeature columns match:",
      list(X_train.columns) == list(X_test.columns))
```

```python
rf_model = RandomForestClassifier(
    n_estimators=200,
    criterion="gini",
    max_depth=None,
    min_samples_split=2,
    min_samples_leaf=1,
    class_weight="balanced_subsample",
    random_state=42,
    n_jobs=-1
)

print("Random Forest model initialized successfully!")
print(rf_model)
```

```python
# Train the Random Forest classifier
rf_model.fit(X_train, y_train)

print("Random Forest training completed successfully!")

print("Number of trees:", len(rf_model.estimators_))
print("Number of disease classes learned:", len(rf_model.classes_))
```

```python
# Generate predictions on test data
y_pred = rf_model.predict(X_test)

print("Predictions generated successfully!")
print("Total predictions:", len(y_pred))

# Compare actual and predicted values
results_df = pd.DataFrame({
    "Actual Disease": y_test.values,
    "Predicted Disease": y_pred
})

display(results_df.head(15))
```

```python
# Calculate evaluation metrics
accuracy = accuracy_score(y_test, y_pred)

precision = precision_score(
    y_test, y_pred,
    average="weighted",
    zero_division=0
)

recall = recall_score(
    y_test, y_pred,
    average="weighted",
    zero_division=0
)

f1 = f1_score(
    y_test, y_pred,
    average="weighted",
    zero_division=0
)

print("RANDOM FOREST MODEL EVALUATION")
print("--------------------------------")
print(f"Accuracy  : {accuracy:.4f}")
print(f"Precision : {precision:.4f}")
print(f"Recall    : {recall:.4f}")
print(f"F1-Score  : {f1:.4f}")
```

```python
print(
    classification_report(
        y_test,
        y_pred,
        zero_division=0
    )
)
```

```python
evaluation_results = pd.DataFrame({
    "Model": ["Random Forest"],
    "Accuracy": [accuracy],
    "Precision (Weighted)": [precision],
    "Recall (Weighted)": [recall],
    "F1-Score (Weighted)": [f1]
})

display(evaluation_results.round(4))
```

```python
from pathlib import Path

evaluation_path = Path("../evaluation")
evaluation_path.mkdir(parents=True, exist_ok=True)

evaluation_results.to_csv(
    evaluation_path / "random_forest_metrics.csv",
    index=False
)

print("Random Forest evaluation metrics saved successfully!")
```

```python
import joblib
from pathlib import Path

model_path = Path("../models")
model_path.mkdir(parents=True, exist_ok=True)

joblib.dump(
    rf_model,
    model_path / "random_forest_model.pkl"
)

print("Random Forest model saved successfully!")
```

```python
import json

model_config = {
    "model": "Random Forest Classifier",
    "n_estimators": 200,
    "criterion": "gini",
    "max_depth": None,
    "min_samples_split": 2,
    "min_samples_leaf": 1,
    "class_weight": "balanced_subsample",
    "random_state": 42,
    "training_records": len(X_train),
    "testing_records": len(X_test),
    "training_disease_classes": int(y_train.nunique()),
    "features": list(X_train.columns),
    "target": "Disease"
}

with open(model_path / "random_forest_config.json", "w") as file:
    json.dump(model_config, file, indent=4)

print("Model configuration saved successfully!")
```
